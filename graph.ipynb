```python
import networkx as nx

# KG用に「有向グラフ」を作成
# (MultiDiGraphは、同じノード間に複数の関係を持てるグラフ)
G = nx.MultiDiGraph()

# --- トリプル構造 (h, r, t) の定義 ---

# 例1： (Python, is_related_to, プログラミング言語)
h1 = 'Python'
r1 = 'is_related_to'
t1 = 'programming language'

# グラフに追加:
# h1(Python) から t1(プログラミング言語) へ矢印を引き、
# その矢印(エッジ)に r1(is_related_to) というラベルを付けます。
G.add_edge(h1, t1, label=r1)

# 例2： (Python, has, オブジェクト指向)
h2 = 'Python'
r2 = 'has'
t2 = 'object oriented'

# グラフに追加:
# h2(Python) から t2(オブジェクト指向) へ矢印を引き、
# その矢印に r2(has) というラベルを付けます。
G.add_edge(h2, t2, label=r2)


# --- グラフの内容確認 ---

# グラフに含まれるノード（h と t）
print(list(G.nodes()))
# 出力: ['Python', 'プログラミング言語', 'オブジェクト指向']

# グラフに含まれるエッジ（h, t, r の関係）
for u, v, data in G.edges(data=True):
    print(f"  (h: {u}, t: {v}, r: {data['label']})")
# 出力:
#   (h: Python, t: プログラミング言語, r: is_related_to)
#   (h: Python, t: オブジェクト指向, r: has)
```

```python
# グラフの可視化
import matplotlib.pyplot as plt
pos = nx.spring_layout(G)  # ノードの配置を決定
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2000, font_size=10)
edge_labels = nx.get_edge_attributes(G, 'label')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='red')
plt.show()
```
